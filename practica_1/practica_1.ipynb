## Cuestionario previo

### 1. ¿Qué es una shell y qué relación existe entre sh y bash?

Una shell es un intérprete de comandos que permite al usuario comunicarse con el sistema operativo mediante instrucciones escritas. `sh` es una de las shells tradicionales de Unix, mientras que `bash` es una shell compatible con `sh` que incorpora funciones adicionales.

### 2. ¿Qué es un script de Bash?

Un script de Bash es un archivo de texto que contiene una secuencia de comandos que pueden ejecutarse automáticamente. La diferencia con ejecutar comandos directamente en la terminal es que el script permite almacenar y repetir una secuencia de instrucciones.

### 3. ¿Qué es Jupyter?

Jupyter es un entorno interactivo basado en notebooks. Un notebook puede contener texto, ecuaciones, código y resultados.

### 4. ¿Qué es Markdown?

Markdown es un lenguaje de marcado ligero utilizado para dar formato al texto. En Jupyter permite crear títulos, listas, tablas, enlaces y ecuaciones.

### 5. ¿Cuál es la diferencia entre Git y GitHub?

Git es un sistema de control de versiones que registra los cambios realizados en los archivos de un proyecto. GitHub es una plataforma en línea que permite almacenar y compartir repositorios Git.

#### Matriz de transformación homogénea

La forma general de la matriz de transformación homogénea de un sistema \(j\) respecto a un sistema \(i\) es:

\[
{}^{i}T_{j} =
\begin{bmatrix}
{}^{i}R_{j} & {}^{i}P_{j} \\
0\quad0\quad0 & 1
\end{bmatrix}
\]

donde:

\[
{}^{i}R_{j}
\]

representa la matriz de rotación y

\[
{}^{i}P_{j}
\]

representa el vector de posición.

#### Matriz de rotación respecto al eje z

\[
R_z(\theta)=
\begin{bmatrix}
\cos\theta & -\sin\theta & 0\\
\sin\theta & \cos\theta & 0\\
0 & 0 & 1
\end{bmatrix}
\]

#### Código del script

El siguiente script automatiza la creación de las carpetas y archivos solicitados en la práctica.

```bash
#!/bin/bash

echo "===== INICIO DE PRACTICA 1 ====="

cd "$HOME" || exit

if [ -d "Practica1" ]; then
    echo "La carpeta Practica1 ya existe."
else
    echo "Creando carpeta Practica1..."
    mkdir Practica1
fi

mkdir -p Practica1/Letras
mkdir -p Practica1/Integrantes

touch Practica1/Letras/a.txt
touch Practica1/Letras/b.txt
touch Practica1/Letras/c.txt

touch Practica1/Integrantes/ValeriaCuriel.txt
touch Practica1/Integrantes/EdsonMuñoz.txt

echo "Estructura creada:"

tree Practica1

echo "Eliminando Practica1..."

rm -r "$HOME/Practica1"

echo "Practica1 eliminada correctamente."

echo "===== FIN DE PRACTICA 1 ====="
```


El script comienza mostrando un mensaje de inicio y posteriormente se desplaza al directorio personal del usuario mediante `$HOME`.

Después verifica si existe la carpeta `Practica1`. Si no existe, la crea.

Posteriormente se generan las carpetas `Letras` e `Integrantes`, junto con los archivos solicitados.

El comando `tree` permite visualizar la estructura creada.

Finalmente, se elimina únicamente la carpeta `Practica1` mediante `rm -r`.

## Análisis de resultados

### 1. ¿Qué ventajas ofrece un script de Bash frente a ejecutar manualmente una secuencia de comandos en la terminal?

Un script de Bash permite automatizar una serie de instrucciones que, de otra forma, tendrían que ejecutarse manualmente una por una. Esto reduce errores, ahorra tiempo y facilita repetir el mismo procedimiento cuando sea necesario.

### 2. ¿Qué diferencias existen entre una celda de Markdown y una celda de código en Jupyter?

Una celda de Markdown se utiliza para escribir texto, títulos, listas, enlaces, imágenes y ecuaciones, mientras que una celda de código permite ejecutar instrucciones y mostrar sus resultados dentro del notebook.

### 3. ¿Qué utilidad tiene un repositorio en línea para almacenar scripts, notebooks y documentación de un proyecto de robótica?

Un repositorio en línea permite almacenar y organizar los archivos de un proyecto, mantener un historial de cambios y compartir el trabajo con otros integrantes. Además, facilita la colaboración y el acceso a los archivos desde diferentes equipos.
